## Revenue Over time

```python
monthly_revenue = Sales.set_index('OrderDate')['ExtendedAmount'].resample('ME').sum() #ME: month end
```

```python
fig, ax = plt.subplots(figsize=(12,8))

ax.plot(monthly_revenue.index, monthly_revenue.values)
ax.set_xlabel('Time')
ax.set_ylabel('Revenue')
plt.title('Revenue over time ($)')
plt.show()
```

## Market value by country

```python
Sales['Profit'] = Sales['ExtendedAmount'] - Sales['TotalProductCost'] #the gains

market_value = Sales.groupby('SalesTerritoryCountry')[['ExtendedAmount', 'Profit']].sum()
market_value.columns = ['Revenue', 'Profit'] #ExtendedAmount = revenue

market_value['Orders'] = Sales.groupby('SalesTerritoryCountry')['SalesOrderNumber'].nunique() #salesorders aren't unique not because it's duplicated in a way we know, a Salesorder in this datsets may have multiple products
market_value['MarginPct'] = market_value['Profit'] / market_value['Revenue'] * 100 #margin

market_value
```

```python
market_value = market_value.sort_values('Revenue', ascending=False)

plt.bar(market_value.index, market_value['Revenue'])
plt.title('Market Value by Country')
plt.ylabel('Revenue ($)')
plt.xticks(rotation=45)
plt.show()
```
