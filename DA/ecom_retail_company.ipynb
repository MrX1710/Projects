```python
Products_csv = pd.read_csv('Products.csv')
customers_json = pd.read_json('customers.json')
internet_sales_excel = pd.read_excel('InternetSales.xlsx', engine='calamine')
sales_territ_parquet = pd.read_parquet('SalesTerritory.parquet')
```

```python
def bold_green(text):
    return f'\033[1;32m{text}\033[0m'
def bold(text):
    return f'\033[1m{text}\033[0m'
```

```python
datasets = {
    "Products (CSV)": Products_csv,
    "Customers (JSON)": customers_json,
    "InternetSales (xlsx)":internet_sales_excel,
    "SalesTerritory":sales_territ_parquet,
}
```

```python
for n, d in datasets.items():
    print(f"name: {bold_green(n)}:\n{d.columns}")
```

```python
##datasets infos
for name, data in datasets.items():
    na_ratio = (data.isna().sum() / len(data)) * 100
    dup_ratio = (data.duplicated().sum() / len(data)) * 100
    print(f'{bold('Dataset')}: {bold_green(name)}:')
    print(f"{bold('Lenght')}: {len(data)}")
    print(f"{bold('Number of columns')}: {len(data.columns)}")
    print(f"{bold('Missing values %')}:\n{(na_ratio[na_ratio > 0])}\n")
    print(f"{bold('Duplicated values %')}:\n{(dup_ratio[dup_ratio > 0])}\n")
```

```python
for name, data in datasets.items():
    print(data.info())
```
