## Internet Sales

the only issue with iternet sales is the currency code, it contains no other nans

```python
#since we have the code of each number we can structure it

Currency = pd.DataFrame({
    'Key'      :          [6,     19,    29,    36,   98,   100],
    'Code'     :          ['AUD', 'CAD', 'DEM', 'EUR', 'GBP', 'USD'],
    'Currency' :          ['Australian Dollar', 'Canadian Dollar', 'Deutsche Mark',
                               'Euro', 'UK Pound', 'US Dollar'],
})
```

```python
# Internet_Sales = Internet_Sales.drop(columns=[6,     19,    29,    36,   98,   100])
```

```python
# Internet_Sales['Key'] = np.nan
```

```python
# Internet_Sales.merge(Currency, on='Key')
```

failure attempts

```python
cols = [29, 6, 98, 19, 36, 100]
Internet_Sales['Key'] = np.nan

for c in cols: #ai
    Internet_Sales.loc[Internet_Sales[c].isna() == False, 'Key'] = c

Internet_Sales = Internet_Sales.drop(columns=cols)
Internet_Sales = Internet_Sales.merge(Currency, on='Key', how='left')
```

```python
# now that every dataset is ready, we must merge them all into one sales dataset
```

```python
Sales = (
    Internet_Sales.merge(
        Products, on='ProductKey')
    .merge(Customers, on='CustomerKey')
    .merge(Sales_Territory, on='SalesTerritoryKey')
)

Sales.to_parquet('Sales.parquet', index=False)
```
